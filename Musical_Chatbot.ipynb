```python
# Simple chatbot program

precepts = ["hello", "tea", "cafe", "no", "thank you", "goodbye"]
# Define a list of known questions and their responses

responses = [
    "Hello! How can I assist you today?",
    "Sure, where would you like to it? A cafe in the campus maybe?",
    "There are 2 cafes - in main building and near the lecture halls.",
    "Is there anything else I can help you with",
    "You are welcome.",
    "Goodbye! Have a great day!"
]
```

```python
# Function to get user input and respond
def chatbot():

    # User welcome and termination instruction.
    print("Chatbot: Hello! How can I assist you today?")
    # Main loop:
    while True:
        
        # 1. Get user input
        user_input = input("You: " )       
        # 2. Check if the user wants to exit
        if user_input == ("bye"):
            print("Chatbot: Thankyou for your interaction.")
            break
            
        found = False 
        # 3. Search for a pattern in the precepts. And fetch a predefined responses
        #   3.1 Check if any of precepts are there in the input
        #   3.2 Return the index
        #   3.3 Use index to find repsonse
        #   3.4 Break the loop
        for pattern in precepts:
            if pattern in user_input:
                ind = precepts.index(pattern)
                response = responses[ind]
                print("Chatbot:" , response)
                found = True
                break;
        
        if not found:
            print("Chatbot: I'm not sure how to respond to that.")
        
        
# Run the chatbot
chatbot()
```

### 2.1 Mapping inputs to responses using dictionaries.

We now use dictionaries to map user intents. Instead of writing a new *dictionary*, use the **ZIP** function on a list. 

```python
# Write a function that takes in precept categories and responses lists and returns a mapping between them.

# - Use ZIP function to combine two independent lists (precepts and responses) and create a new map between them.
# - Search on the ZIP function and how to use it in a for loop
# - Run the main loop and check that this function works

# ___ Write your code here _____

# def zip_responses(precepts, response):

#     mapping = {}
#     return mapping

def zip_responses(precepts, responses):
    mapping = dict(zip(precepts, responses))
    return mapping
```

```python
# Write a function to create precept-response pairs and return them as a Dictionary.
# Give the function a characteristic name (e.g., pattern_responses) 
# You can directly write these precept response pairs or try the ZIP function

# # ___ Write your code here _____
# def mappings():
#     precepts = []
#     responses = []
#     return zip_response(precepts, responses)

def pattern_responses():
    """
    Returns a dictionary of precept-response pairs using zip_responses.
    """
    precepts = [
        "hello",
        "how are you",
        "what is your name",
        "bye"
    ]
    
    responses = [
        "Hi there!",
        "I'm just a bot, but I'm doing fine!",
        "I'm your friendly chatbot.",
        "Goodbye!"
    ]
    
    return zip_responses(precepts, responses)

    
def main():
    response_map = pattern_responses()
    
    print("Chatbot: Hello! Type something (type 'bye' to exit).")
    while True:
        user_input = input("You: ").lower()
        if user_input in response_map:
            print("Chatbot:", response_map[user_input])
        else:
            print("Chatbot: Sorry, I don't understand.")
        if user_input == "bye":
            break

# Run the main loop
main()
```

- Match complete words using regular expressions.

```python
# Global state to keep user-specific info
user_state = {
    'name': 'there',
    'color': None
}
```

```python
# Write a function to create 5 RE patterns and return them as separate variables.
# Give the function a characteristic name (e.g., re_patterns) 
# ___ Write your code here _____
def re_patterns():
    greetings = r"\b(hi|hello|hey|hola|howdy)\b"
    farewells = r"\b(bye|goodbye|see you|take care|later)\b"
    name_queries = r"\b(what is your name|who are you|your name|tell me your name|name please)\b"
    wellbeing = r"\b(how are you|how's it going|how do you do|are you okay|you good)\b"
    thanks = r"\b(thanks|thank you|cheers|much appreciated|gracias)\b"
    return greetings, farewells, name_queries, wellbeing, thanks
```

You will write functions for main loop, detecting patterns and generating responses to make chatbot code reusable.

```python
## Detect pattern - Write a function to detect RE patterns in user input and return response category.

## Create the detect_pattern function with one argument : user_input
## 1. Call the re_patterns function and store patterns in seperate variables
## 2. Write if-else conditions to check for pattern match
##    2.1 For the if condition, use the re.search to check if the pattern exists in user input 
##    2.2 If there is a match, return the response type (e.g., "greetings" or "bye" or "happy")
##    2.3 Note: Use <span style="color:red; font-size: 14px"> elif </span> to match multiple conditions
##    2.4 If none of the conditions, return "default" which will be the default response of the chatbot
##    2.5 <span style="color:red; font-size: 14px"> Note: use else for last conditon </span>


# Give the function a characteristic name (e.g., detect_pattern) 
# ___ Write your code here _____
def detect_pattern(user_input):
    global user_state
    greetings, farewells, name_queries, wellbeing, thanks = re_patterns()

    if re.search(greetings, user_input, re.IGNORECASE):
        return "greetings"
    elif re.search(farewells, user_input, re.IGNORECASE):
        return "farewells"
    elif re.search(name_queries, user_input, re.IGNORECASE):
        return "name"
    elif re.search(wellbeing, user_input, re.IGNORECASE):
        return "wellbeing"
    elif re.search(thanks, user_input, re.IGNORECASE):
        return "thanks"

    # Advanced pattern: Name input
    name_input_pattern = r"(?:name is|I'm|i am|my name is) ([A-Z][a-z]*)"
    name_match = re.search(name_input_pattern, user_input,  re.IGNORECASE)
    if name_match:
        user_state['name'] = name_match.group(1)
        return "name_response"

    # Advanced pattern: Color input
    color_input_pattern = r"(?:color is|I like) (\w+)"
    color_match = re.search(color_input_pattern, user_input, re.IGNORECASE)
    if color_match:
        user_state['color'] = color_match.group(1)
        return "color_response"

    return "default"
```

```python
## Chatbot response - Write a function to return a responses by indexing dictionary with response categories.

## Create the chatbot_response function with one argument : user_input
## 1. Call the detect_pattern function and store the response in variable (e.g., detected_type)
## 2. Call the re_responses function and store responses in a variable (e.g., re_response)
## 2. Write if-else conditions to check for pattern match
##    2.1 For the if condition, check if the detected_type is one of the keys of the dictionary
##    2.2 If there is a match, return the response from dictonary
##    2.3 Note: Use elif to match multiple conditions
##    2.4 If none of the conditions, use "default" response

# Give the function a characteristic name (e.g., detect_pattern) 
# _ Write your code here ___
def re_responses():
    return {
        "greetings": "Hello there! How can I assist you?",
        "farewells": "Goodbye! Have a great day!",
        "name": "I'm a simple chatbot created to help you!",
        "wellbeing": "I'm doing great, thanks for asking! How about you?",
        "thanks": "You're welcome!",
        "name_response": "Hello {name}, how are you today?",
        "color_response": "Great to know that your favourite color is {color}!",
        "default": "I'm not sure I understand. Can you please elaborate?"
    }
```

```python
def chatbot_response(user_input):
    detected_type = detect_pattern(user_input)
    responses = re_responses()
    name = user_state['name']
    color = user_state['color']

    if detected_type in responses:
        response = responses[detected_type]
        if detected_type == "name_response":
            return response.format(name=name)
        elif detected_type == "color_response":
            return response.format(color=color)
        else:
            return response
    else:
        return responses["default"]
```

```python
## A reminder for what the main loop looked like.
def chat():
    print("Chatbot: Hello! How can I help you today?")
    while True:
        user_input = input("You: ")
        
        if user_input.lower() == "bye":
            print("Chatbot: Thanks for your interaction. Goodbye!")
            break

        bot_response = chatbot_response(user_input)
        print(f"Chatbot: {bot_response}")

# Start chatting with the bot
chat()
```

### 3.2 Creating mappings from patterns to intents and responses

The logical choice for creating these mappings is to use dictionaries. 

**Remember**: While constructing a dictionary, you want the the **key-value** to be same data type. But, patterns are regular expressions, intents are strings, and responses are lists or strings.

#### Strings to regular expressions

```python
# To convert patterns from strings to REs, we have to look 
# at each pattern and concatenate them with a pipe symbol.

# Loop thorugh patterns list and tags.
 # For each pattern in list, add an "r" symbol
 # Create a dictionary with pattern and tags

# patterns = ["hello","hi","hiya"]

# r"hello|hi|hiya"

# r"|".join(patterns)

def create_pattern_to_intent_mapping(intents_json):
    rgx2int = {}
    for intent in intents_json['intents']:
        patterns = intent['patterns']
        # Join all patterns with the pipe symbol to form a regex
        pattern = r"|".join(patterns)
        rgx2int[pattern] = intent['tag']
    return rgx2int
```

#### Create the dictionary from patterns to tags

```python
## Create an empty dictionary variable rgx2int for mapping from pattern to tags/intents

## Write code here....
int2res = {}
```

#### Create the dictionary from tags to responses


TIP: For any feature, write this as a function. (e.g. storeResponses)

```python
## Create an empty dictionary int2res for mapping from intent to responses.

# Loop through the list of JSON objects:
 # For each object in list, get responses and tag
 # Add to the dictionary: tags as keys and responses as values

def create_intent_to_response_mapping(intents_json):
    
    for intent in intents_json['intents']:
        tag = intent['tag']
        responses = intent['responses']
        int2res[tag] = responses
    return int2res

# Create the mappings from patterns to intents and intents to responses

int2res = create_intent_to_response_mapping(intents_json)
rgx2int = create_pattern_to_intent_mapping(intents_json)

# Check the mappings
print(rgx2int)
print(int2res)
```

Initializing stemmer and lemmatizer for word processing ⬇️

```python
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
```

Tokenizes and preprocesses the input text by converting it to lowercase,removing stopwords, and performing stemming and lemmatization ⬇️

```python
def preprocess_input(user_input, debug=False):

    tokens = word_tokenize(user_input.lower())
    
    stop_words = set(stopwords.words('english'))
    tokens = [word for word in tokens if word not in stop_words]
    
    tokens = [word for word in tokens if word.isalnum()]
    
    stemmed_words = [stemmer.stem(word) for word in tokens]
    lemmatized_words = [lemmatizer.lemmatize(word) for word in tokens]
    
    if debug:
        print("Tokens after preprocessing:")
        print("Stemmed words:", stemmed_words)
        print("Lemmatized words:", lemmatized_words)
    
    return stemmed_words, lemmatized_words
```

Generates a response based on the user's mood ⬇️

```python
def generate_mood_response(mood, user_name):
    for intent in intents_json['intents']:
        if intent['tag'] == mood:
            response = random.choice(intent['responses'])  
            return response.replace("{user_name}", user_name)
            break
    return "I'm not sure how to respond to that mood."
```

Detects the best matching intent from the user's input by comparing it with predefined patterns ⬇️

```python
def detect_pattern(user_input):
    # Preprocess user input and show tokens
    stemmed_words, lemmatized_words = preprocess_input(user_input, debug=False)
    user_tokens = set(lemmatized_words)
    
    best_match = None
    best_score = 0
    
    for intent in intents_json["intents"]:
        for pattern in intent["patterns"]:
            _, pattern_lemmas = preprocess_input(pattern, debug=False)
            pattern_tokens = set(pattern_lemmas)
    
            if not pattern_tokens:
                continue
    
            overlap = len(user_tokens & pattern_tokens) / len(pattern_tokens)
    
            if overlap > best_score:
                best_score = overlap
                best_match = intent["tag"]
    
    if best_score >= 0.5:
        return best_match
    return "default"
```

Main chatbot function that runs the conversation loop, asking the user for their name, detecting their mood, and providing responses accordingly ⬇️

```python
def chatbot():
    print(Fore.BLUE + "Chatbot: Hello! What is your name?" + Style.RESET_ALL)
    user_name = input(Fore.CYAN + "You: " + Style.RESET_ALL)
    
    print(Fore.BLUE + f"Chatbot: Hi, {user_name}! I'm Musical Chatbot" + Style.RESET_ALL)
    print(Fore.BLUE + f"Chatbot: {user_name}! How are you feeling today?" + Style.RESET_ALL)
    
    while True:
        user_input = input(Fore.CYAN + "You: " + Style.RESET_ALL)
        
        # If the user asks for a joke
        if "joke" in user_input.lower():
            joke = get_joke()  # Get a random joke
            print(Fore.BLUE + f"Chatbot: Here's a joke for you, {user_name}: {joke}" + Style.RESET_ALL)
            continue  # Skip the rest of the flow and go back to the user input
        
        # Check if a mood is detected first
        mood = detect_mood(user_input)
        if mood != "mood_neutral":  # If a mood is detected
            response = generate_mood_response(mood, user_name)
            print(Fore.BLUE + f"Chatbot: {response}" + Style.RESET_ALL)
            continue  # Skip the rest of the code, go back to the next user input
        
        # If no mood is detected, proceed to the regular intent-based response
        detected_type = detect_pattern(user_input)
        
        if detected_type == "goodbye":
            print(Fore.BLUE + f"Chatbot: {generate_response(user_input, user_name)}, Goodbye!" + Style.RESET_ALL)
            break
        
        # Generate and print the chatbot's response based on intent
        response = generate_response(user_input, user_name)
        
        # If no response is generated, it means the chatbot didn't understand
        if response:
            print(Fore.BLUE + f"Chatbot: {response}" + Style.RESET_ALL)
        else:
            print(Fore.BLUE + "Chatbot: I'm sorry, I couldn't understand that." + Style.RESET_ALL)
            
            

# Run the chatbot
chatbot()
```
